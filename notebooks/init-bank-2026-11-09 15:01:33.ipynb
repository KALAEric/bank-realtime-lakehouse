```python
%sql
CREATE CATALOG IF NOT EXISTS bank;
CREATE SCHEMA IF NOT EXISTS bank.bronze;
CREATE SCHEMA IF NOT EXISTS bank.silver;
CREATE SCHEMA IF NOT EXISTS bank.gold;
CREATE SCHEMA IF NOT EXISTS bank.ops;
CREATE VOLUME IF NOT EXISTS bank.ops.landing;
CREATE VOLUME IF NOT EXISTS bank.ops.checkpoints;
```

```python
%sql
CREATE TABLE bank.ops.sandbox (id INT, name STRING, amount DECIMAL(10,2));
INSERT INTO bank.ops.sandbox VALUES (1,'Alice', 100.00),(2, 'Bob',250.00), (3, 'Chloé', 75.00);
UPDATE bank.ops.sandbox SET amount = 300.00 WHERE id = 2;
DELETE FROM bank.ops.sandbox WHERE id = 3;
```

```python
%sql
DESCRIBE HISTORY bank.ops.sandbox;
DESCRIBE DETAIL bank.ops.sandbox;

DESCRIBE EXTENDED bank.ops.sandbox;

SELECT * FROM bank.ops.sandbox VERSION AS OF 1; 
RESTORE TABLE bank.ops.sandbox TO VERSION AS OF 1;
DESCRIBE HISTORY bank.ops.sandbox;

```

```python
from delta.tables import DeltaTable

dt=DeltaTable.forName(spark, "bank.ops.sandbox")

display(dt.history().select("version", "timestamp", "operation", "operationParameters", "operationMetrics"))

v1=spark.read.option("versionAsOf", 1).table("bank.ops.sandbox")
display(v1)
```
